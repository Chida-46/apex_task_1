```python
import pandas as pd
import numpy as np
from pathlib import Path

print("Libraries imported successfully!")
```

```python
file_path = Path("ApexPlanet_DataAnalytics_Dataset.xlsx")

df = pd.read_excel(
    file_path,
    sheet_name="Sales_Dataset"
)

print("Dataset loaded successfully!")
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

display(df.head())
```

```python
# Display all column names
print("Column Names:")
print(df.columns.tolist())

# Check data types and non-null counts
print("\nDataset Information:")
df.info()

# View statistical summary
print("\nStatistical Summary:")
display(df.describe(include="all").T)
```

```python
missing_values = df.isnull().sum()

print("Missing values in each column:")
display(missing_values.to_frame(name="Missing Count"))

print("Total missing cells:", df.isnull().sum().sum())
```

```python
# Exact duplicate rows
print("Exact duplicate rows:", df.duplicated().sum())

# Repeated order IDs
print(
    "Repeated Order_ID rows:",
    df["Order_ID"].duplicated().sum()
)

# Count unique order IDs
print("Unique Order_IDs:", df["Order_ID"].nunique())
```

```python
print("Order_Date data type:", df["Order_Date"].dtype)

print("Earliest order date:", df["Order_Date"].min())
print("Latest order date:", df["Order_Date"].max())

display(df["Order_Date"].head(10))
```

```python
categorical_columns = [
    "Gender", "City", "Product", "Category"
]

for col in categorical_columns:
    print(f"\n--- {col} ---")
    print(df[col].value_counts(dropna=False))
```

```python
expected_sales = (
    df["Quantity"] * df["Unit_Price"]
).round(2)

sales_mismatch = (
    df["Total_Sales"].round(2) != expected_sales
)

print("Sales calculation mismatches:", sales_mismatch.sum())

if sales_mismatch.any():
    display(df.loc[
        sales_mismatch,
        ["Order_ID", "Quantity", "Unit_Price", "Total_Sales"]
    ].head(20))
```

```python
# Find all rows with repeated Order_ID values
repeated_orders = df[
    df["Order_ID"].duplicated(keep=False)
].sort_values("Order_ID")

display(repeated_orders)

# Check whether repeated IDs have different products or details
summary = (
    repeated_orders.groupby("Order_ID")
    .agg(
        Rows=("Order_ID", "size"),
        Customers=("Customer_ID", "nunique"),
        Dates=("Order_Date", "nunique"),
        Products=("Product", "nunique"),
        Total_Sales=("Total_Sales", "sum")
    )
)

display(summary)
```

```python
clean_df = df.copy()

print("Original shape:", df.shape)
print("Working copy shape:", clean_df.shape)
```

```python
clean_df["Order_Date"] = pd.to_datetime(
    clean_df["Order_Date"],
    errors="coerce"
)

print("New date type:", clean_df["Order_Date"].dtype)
print("Invalid dates:", clean_df["Order_Date"].isna().sum())
```

```python
# Record the original missing counts
print("Missing Age before:", clean_df["Age"].isna().sum())
print("Missing City before:", clean_df["City"].isna().sum())

# Fill missing ages with the median
age_median = clean_df["Age"].median()
clean_df["Age"] = clean_df["Age"].fillna(age_median)

# Preserve missing city information explicitly
clean_df["City"] = clean_df["City"].fillna("Unknown")

print("\nMissing values after treatment:")
display(clean_df.isnull().sum().to_frame("Missing_Count"))
```

```python
# Check invalid ages
invalid_ages = clean_df[
    (clean_df["Age"] < 0) |
    (clean_df["Age"] > 120)
]

# Check invalid quantities and prices
invalid_quantity = clean_df[clean_df["Quantity"] <= 0]
invalid_price = clean_df[clean_df["Unit_Price"] <= 0]

print("Invalid ages:", len(invalid_ages))
print("Non-positive quantities:", len(invalid_quantity))
print("Non-positive prices:", len(invalid_price))
```

```python
text_columns = [
    "Order_ID",
    "Customer_ID",
    "Customer_Name",
    "Gender",
    "City",
    "Product",
    "Category"
]

for col in text_columns:
    clean_df[col] = clean_df[col].astype("string").str.strip()

print("Text fields standardized.")
```

```python
expected_sales = (
    clean_df["Quantity"] * clean_df["Unit_Price"]
).round(2)

sales_mismatch = (
    clean_df["Total_Sales"].round(2) != expected_sales
)

print("Sales calculation mismatches:", sales_mismatch.sum())
```

```python
# Extract month and year from the date
clean_df["Sales_Month"] = clean_df["Order_Date"].dt.month_name()
clean_df["Sales_Year"] = clean_df["Order_Date"].dt.year

# Group customers by age
clean_df["Customer_Age_Group"] = pd.cut(
    clean_df["Age"],
    bins=[0, 25, 35, 50, 65, float("inf")],
    labels=["18-25", "26-35", "36-50", "51-65", "66+"]
)

display(clean_df.head())
print("Updated shape:", clean_df.shape)
```

```python
from pathlib import Path

output_dir = Path("output")
output_dir.mkdir(exist_ok=True)

clean_df.to_csv(
    output_dir / "cleaned_sales_dataset.csv",
    index=False
)

clean_df.to_excel(
    output_dir / "cleaned_sales_dataset.xlsx",
    index=False
)

print("Cleaned files saved to:", output_dir.resolve())
```

```python
# Preserve the original IDs for reference
clean_df["Original_Order_ID"] = clean_df["Order_ID"]

# Count each occurrence of an Order_ID
occurrence = clean_df.groupby("Order_ID").cumcount()

# Keep the first ID; assign a unique suffix to later occurrences
duplicate_mask = occurrence > 0

clean_df.loc[duplicate_mask, "Order_ID"] = (
    clean_df.loc[duplicate_mask, "Original_Order_ID"]
    + "_FIX"
    + occurrence[duplicate_mask].astype(str).str.zfill(2)
)

# Validate the correction
print("Rows:", len(clean_df))
print("Unique Order IDs:", clean_df["Order_ID"].nunique())
print("Remaining duplicate Order IDs:", clean_df["Order_ID"].duplicated().sum())

display(
    clean_df[clean_df["Original_Order_ID"].duplicated(keep=False)]
    [["Original_Order_ID", "Order_ID", "Order_Date", "Customer_ID", "Product"]]
    .sort_values("Original_Order_ID")
)
```

```python
from pathlib import Path

output_dir = Path("output")
output_dir.mkdir(exist_ok=True)

clean_df.to_csv(
    output_dir / "cleaned_sales_dataset.csv",
    index=False
)

clean_df.to_excel(
    output_dir / "cleaned_sales_dataset.xlsx",
    index=False
)

print("Updated files saved successfully.")
```

```python
print("Dataset shape:", clean_df.shape)
print("Missing values:", clean_df.isnull().sum().sum())
print("Exact duplicate rows:", clean_df.duplicated().sum())
print("Duplicate Order IDs:", clean_df["Order_ID"].duplicated().sum())

sales_check = (
    clean_df["Total_Sales"]
    - clean_df["Quantity"] * clean_df["Unit_Price"]
).abs()

print("Sales calculation mismatches:", (sales_check > 0.01).sum())
print("Order_Date data type:", clean_df["Order_Date"].dtype)
```
